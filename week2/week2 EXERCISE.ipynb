```python
#imports

import os
from dotenv import load_dotenv
from openai import OpenAI
import gradio as gr
```

```python
load_dotenv(override=True)

openai_api_key = os.getenv('OPENAI_API_KEY')
if openai_api_key:
    print("Your chillin', Good to go")
else:
    print("Something is wrong!")

MODEL = "gpt-4.1-mini"
openai = OpenAI()
```

```python
system_message = """
    You are a Computer Science professor who has many years of 
    experience in most programming languages. Give detailed but short answers.
    Always be accurate. If you don't know the answer, say so
    """

question = """
    Please explain what this code does and why:
    yield from {book.get("author") for book in books if book.get("author")}
"""
```

```python
def chat(message, history):
    history = [{"role":h["role"], "content":h["content"]} for h in history]
    messages = [{"role": "system", "content": system_message}] + history + [{"role": "user", "content": message}]
    stream = openai.chat.completions.create(model=MODEL, messages=messages, stream=True)
    response = ""
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        yield response
```

```python
gr.ChatInterface(fn=chat, 
                 type="messages",
                 title="Week 2 Exercise",
                 examples=[question]).launch()
```
